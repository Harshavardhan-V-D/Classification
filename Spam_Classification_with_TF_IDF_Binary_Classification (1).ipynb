```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
)
```

```python
messages = pd.read_csv('SMSSpamCollection', sep='\t', names = ['label', 'message'])
```

```python
corpus = []
y = pd.get_dummies(messages['label'])
y = y.iloc[:,1].values
for i in range(0, len(messages['message'])):
  corpus.append(messages['message'][i])


X_train, X_test, y_train, y_test = train_test_split(corpus, y, test_size= 0.2, random_state=0)
```

```python
vec = TfidfVectorizer(ngram_range=(1,2), stop_words="english", sublinear_tf=True)
X_train = vec.fit_transform(X_train)
X_test = vec.transform(X_test)
```

```python
clf = LogisticRegression(C=2.0, class_weight="balanced", random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
```

```python
confusion_m = confusion_matrix(y_test, y_pred)
p = precision_score(y_test, y_pred, zero_division=0)
r = recall_score(y_test, y_pred)
print(confusion_m)
print(p)
print(r)
```
